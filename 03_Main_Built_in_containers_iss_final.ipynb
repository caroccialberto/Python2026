## Range

Range is a function that generates sequences of integers. The variable produced can be used as an iterator for other functions:

```python
range(4) #this will produce the elements from 0 (by default) to the number BEFORE the last one indicated
```

```python
my_list = list(range(4)) #we can use it in a list
my_list
```

```python
list(range(3, 7)) #we can specify a starting index
```

```python
list(range(3, 10, 2)) #we can add a third argument: the step (or increment). by default, the step is +1, here we change it to +2
```

### Range

Actually range does not build a list, but a sequence, that is an object iterable in a given direction.

Basically, range returns an object that you can scroll from the first to the last element, one element after another, always in the same direction.

From this object you can build a container, such as a list.

The reason why range does not provide a container directly is that if you have a very long sequence you should store all of them in your memory, which might be useless as well as memory-consuming. **Range stores the algorithm instead** (so called "lazy") and can compute and provide a following item in the sequence one by one in real time; the advantage will be clearer when we will see **for cycles** in control flows.

## Sets

A set is an __unordered__ collection, without duplicates, __mutable__, containing arbitrary values in arbitrary number. We also have the __immutable__ version, the frozenset, which can't be modified once created.
They are not ordered in the sense that the elements are not maintained in a meaningful user order; this because the order used by the interpreter is useful to make fast searching of items.


```python
s = set()
s.add(2)
s.add(3)
s.add(2) # this is a duplicate! it gets ignored and isn't added twice
s.add(4)
s
```

### Sets

```python
u = set(['alfa', 2, 3.5])
u
```

Curly brackets can also be used to declare the variable the `'{}'`. If you want an empty set, you must use s = set(). If you type s = {}, Python creates an empty dictionary, not an empty set!

```python
u = {'alfa', 2, 3.5}
u
```

### Sets can be created from any sequence (python is a very inclusive language)

```python
l=[1, 2, 2, 3, 3, 3, 4, 4, 4, 4] #list
t = set(l) #we can use set to clear duplicates, it is a very unique and important feature! REMEMBER IT!
t
```

```python
l = range(4) #range
t = set(l)
t
```

### Sets

```python
s
```

```python
s.intersection(t) #elements present in both sets
```

```python
s.difference(t)  # elements present in s but not in t!
```

```python
t.difference(s) #elements present in t but not in s!
```

```python
s.union(t) #union of sets (withouth duplicates). if the resulting set is empty it just does nothing (no error)
```

We can also write some of the sets operations in other ways

```python
o = s&t #intersection, just like s.intersection(t)
print (o)
```

## Dictionaries
A dictionary (dict) is a container that associates an arbitrary and __unique key__ to an arbitrary __value__. Think of it like a real-world dictionary:

Key: The word you look up (Must be unique and __immutable__, like a string or number).

Value: The definition (Can be anything: a number, a list, another dictionary).

Basically, the keys are handled as a set, but each key is uniquely associated with a value.

In other languages it is known as an associative array or a map.


### Dicts

```python
atomic_number = {'H': 1, 'He': 2, 'C': 6, 'Fe': 26} #creating directly a dict, using curly braces {} and colons: best for when you know the data beforehand
atomic_number
```

```python
atomic_number['C'] #accessing an element of the dict through its key
```

They are not necessarily homogeneous, meaning they can host different types of key/values, the important thing is that they are PAIRED. So, one key, one value:

```python
d = {}
d['alpha'] = 3 #creating a "space" in the dictionary at the specific key indicated, the variable put at the right of the equation
d[2.5] = 'xyz'
d[3+4j] = [3, 4, 5]
d[(1,2,3)] = { 'x': 2, 'y': 3, 'z': 1 }
#could we write d[[1,2,3]] = 5? and why?
d
```

### Dicts

```python
list(d.keys())
#what does d.keys() produce?
```

```python
list(d.values())
#what does d.values() produce?
```

```python
list(d.items())
#what does d.items() produce?
```

### Accessing dicts

```python
d['alpha'] #standard access to the dict, we know the key is there and we want to retrieve its content
```

```python
'alpha' in d #in fact, whe can print a statement about the existance of the key in the dict
# try also 'beta' in d
```

```python
d.get('beta', -1999)  #but what is happening here? we are asking for a DEFAULT: in other words, we want to get what is at the key 'beta'
#if the key is not present, we are returned with the default specified
#this DOES NOT modify the dict
```

```python
d
```

```python
d.setdefault('alpha', -1999)
d.setdefault('beta', -1999)
#differently from the .get() function, the setdefault() ADDS the indicated key in the dictionary, modifying it, with the default value if the key is not present
```
