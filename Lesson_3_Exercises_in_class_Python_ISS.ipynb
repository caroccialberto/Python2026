# Ex.1 Print with conditional
Print the elements of the given list if:
-  they are strings
- start with the letter "A" or "a"
- The length of the string is more than 3

Store in result_list all elements that are not printed

```python
#@title utils
def check_result(result_list, expected_solution):
  if result_list == expected_solution:
    print("OK, TEST PASSED")
  else:
    print("NOT OK")
```

```python
given_list = ["Andrea","A",3,[1,2,3,4,5,6],"Luigi","and","asdasd"]

```

```python

def print_conditioned(given_list):
  result_list =[]
  print(result_list)
  for i in given_list:
    if type(i) == str and i[0] in ["a", "A"] and len(i) > 3:
      print(i)
    else:
      result_list.append(i)
  print(result_list)

print_conditioned(given_list)
```

#Ex.2 Count even and odds
Given a list of integers, count how many odd and how many even numbers there are
- Store results in the list result_list `[n_odd,n_even]` if `n_odd > n_even` and `[n_even, n_odd]` otherwise


```python
list_of_integers = [1,4,566,123,44,9932,11]
```

```python
def count_odd_even(list_of_integers):
  n = {"n_odd" : 0, "n_even" : 0}
  for i in list_of_integers:
    if i%2 == 0:
      n["n_even"] += 1
    else:
      n["n_odd"] += 1
  n_list = list(n.values())
  if n_list[0] > n_list[1]:
    print([n_list[0], n_list[1]])
  else:
    print([n_list[1],n_list[0]])

count_odd_even(list_of_integers)
```

```python
result_list = count_odd_even(list_of_integers)
print(result_list)
expected_solution = [4,3]
check_result(result_list, expected_solution)
```

#Ex.3 Concatenate integers
given int1 and int2 , retun the concatenated integer that is larger
es. int1 = 34, int2 = 9 --> return 934
es. int1 = 21, int2 = 123 --> return 21123

```python
int1 = 981
int2 = 98

def concatenate_integers(int1, int2):
  s_int1 = str(int1)
  s_int2 = str(int2)
  if len(s_int1) <= len(s_int2):
    for i in range(len(s_int1)):
      if s_int1 == s_int2:
        print(int(str(int1)+str(int2)))
        break
      elif int(s_int1[i]) > int(s_int2[i]):
        print(int(s_int1+s_int2))
        break
      elif int(s_int1[i]) <= int(s_int2[i]):
        print(int(s_int2+s_int1))
        break
  else:
      for i in range(len(s_int2)):
        if int(s_int1[i]) > int(s_int2[i]):
          print(int(s_int1+s_int2))
          break
        elif int(s_int1[i]) <= int(s_int2[i]):
          print(int(s_int2+s_int1))
          break
```

```python
concatenate_integers(int1, int2)
```
