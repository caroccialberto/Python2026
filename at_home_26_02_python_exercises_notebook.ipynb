## Topic 1 — Basic Python (lists, dictionaries, functions, loops, if/else)

### Brief recap
In basic Python, you combine **variables**, **lists**, and **dictionaries** with **control flow** (`if/elif/else`, `for`, `while`) and **functions** (`def`) to build reusable logic.

Core ideas:
- **Lists** store ordered collections (`items[0]`, slicing, append, loop over elements).

- **Functions** help you reuse code and make programs easier to test and read.

### Exercise 1.1 (Easy) — Classify numbers with `if/else` and loops

Given the list `numbers`, create three lists:
- `negatives`
- `zeros`
- `positives`

Then print:
1. the three lists
2. how many values are in each list

Use a `for` loop and `if/elif/else`.

```python
numbers = [4, -3, 0, 8, -1, 0, 7, -9, 2]

# TODO: create the three lists
negatives = []
zeros = []
positives = []

# TODO: loop through numbers and classify each value
for x in numbers:
  if x<0:
    negatives.append(x)
  elif x > 0:
    positives.append(x)
  elif x == 0:
    zeros.append(x)
# TODO: print the results
print(f"negatives: {negatives}")
print(f"zeros: {zeros}")
print(f"positives: {positives}")
```

### Exercise 1.3 (Easy) — Write and use a function

Write a function `grade_label(score)` that returns:
- `"A"` if score >= 90
- `"B"` if score >= 75
- `"C"` if score >= 60
- `"D"` otherwise

Then apply it to all values in `scores` and store the labels in a list.

```python
scores = [95, 81, 67, 59, 73, 88]

def grade_label(score):
    # TODO: implement with if/elif/else
    if score >= 90:
      return "A"
    elif score>=75:
      return "B"
    elif score >=60:
      return "C"
    else:
      return "D"

# TODO: create a list of labels by calling grade_label on each score
labels = [grade_label(x) for x in scores]

# TODO: print scores and labels
print(labels)
```

### Exercise 2.4 (Intermediate) — Points on circumferences with `scatter`

Create a plot with points sampled on the circumferences of multiple circles.

Given:
- centers `(cx, cy)` for each circle
- a radius for each circle
- number of points per circle (`n_points`)

Tasks:
1. Use NumPy to generate angles `theta` from `0` to `2π`.
2. For each circle, compute the points:
   - `x = cx + r*cos(theta)`
   - `y = cy + r*sin(theta)`
3. Plot the points with `matplotlib.pyplot.scatter`.
4. Use equal aspect ratio so circles look like circles.
5. Add title and grid.

You may use either a loop over circles or broadcasting.

```python
import numpy as np
import matplotlib.pyplot as plt
import math

centers = np.array([
    [0.0, 0.0],
    [2.5, 1.0],
    [-2.0, 2.0]
])
radii = np.array([1.0, 0.7, 1.4])
n_points = 10000

# TODO: create theta values (angles)
rng = np.random.default_rng(42)
theta = rng.uniform(0, 2*np.pi, n_points)



# TODO: create the scatter plot
plt.figure(figsize=(6, 6))

# Example structure (complete it):
for (cx, cy), r in zip(centers, radii):
    x = cx + r*np.cos(theta)
    y = cy + r*np.sin(theta)
    plt.scatter(x, y, s=10, label=f"r={r}")

# TODO: formatting
plt.title("my plot")
plt.axis("equal")
plt.grid(True)
plt.legend()
plt.show()
```

### Exercise 3.2 (Easy) — Histograms of two distributions on the same axis

Generate two distributions and plot their histograms on the same axes.

Tasks:
1. Use `np.random.seed(0)` for reproducibility.
2. Generate:
   - `a`: 1000 samples from a normal distribution (mean 0, std 1)
   - `b`: 1000 samples from a normal distribution (mean 2, std 0.5)
3. Plot both histograms on the same axis with transparency (`alpha=...`).
4. Add legend, title, and axis labels.

```python
import numpy as np
import matplotlib.pyplot as plt

# TODO: set seed and generate the two distributions
np.random.seed(0)
n = 100000
a = np.random.normal(size = n)
b = np.random.normal(loc = 2, scale = 0.5, size = n)

plt.figure(figsize=(7, 4))

# TODO: plot both histograms on the same axis
plt.hist(a, bins=30, density= True, alpha=0.6, label="A")
plt.hist(b, bins=30, density= True, alpha=0.6, label="B")

# TODO: formatting
plt.legend()
plt.title("Hist of Normals")
plt.xlabel("Freq")
plt.ylabel("Probability")
plt.show()
```
