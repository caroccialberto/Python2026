```python
n = 1000000
c = 0
r = 1.
for i in range(n):
  x = random.uniform(0,1.)
  y = random.uniform(0,1.)
  d = (x**2 + y**2)**0.5
  if d <= r:
    c += 1
```

```python

P_a = c/n
a = P_a*r*2*2
pi = a*r**2

print(P_a)
print(pi)
```

```python
def MC_pi(n = 10000, r = 1.):
  c = 0
  for i in range(n):
    x = random.uniform(0,r)
    y = random.uniform(0,r)
    d = (x**2 + y**2)**0.5
    if d <= r:
      c += 1
  P_a = c/n
  pi = P_a*4
  return P_a, pi
```

```python
P_a, pi = MC_pi(n = 1000000, r = 3.)
print(P_a)
print(pi)
```
